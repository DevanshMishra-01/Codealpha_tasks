```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Theme
sns.set_theme(style="whitegrid", palette="Set2")

# Load dataset
df = pd.read_csv("odi_clean_dataset.csv")

# Convert numeric columns
cols = ['bat_Mat','bat_Runs','bat_Ave','bat_SR','bat_100','bat_50',
        'bowl_Mat','bowl_Wkts','bowl_Ave','bowl_Econ','bowl_SR']

for col in cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

## TOP 10 RUN SCORERS

```python
##This shows the top 10 run scorers in odi format

top_runs = df.nlargest(10, "bat_Runs")

plt.figure(figsize=(10,6))
sns.barplot(data=top_runs,
            x="bat_Runs",
            y="Player",
            hue="Player",
            legend=False)

plt.title("Top 10 Run Scorers")
plt.xlabel("Runs")
plt.ylabel("Player")
plt.show()
```

## TOP 10 WICKET TAKERS

```python
## consists the details about the 10 top wicket takers

top_wickets = df.nlargest(10, "bowl_Wkts")

plt.figure(figsize=(10,6))
sns.barplot(data=top_wickets,
            x="bowl_Wkts",
            y="Player",
            hue="Player",
            legend=False)

plt.title("Top 10 Wicket Takers")
plt.xlabel("Wickets")
plt.ylabel("Player")
plt.show()
```

```python
## Batting average distribution 

plt.figure(figsize=(8,5))

sns.histplot(df["bat_Ave"],
             bins=20,
             kde=True)

plt.title("Batting Average Distribution")
plt.show()
```

## Strike rate distribution

```python
plt.figure(figsize=(8,5))

sns.histplot(df["bat_SR"],
             bins=20,
             kde=True)

plt.title("Batting Strike Rate Distribution")
plt.show()
```

## Runs vs Matches

```python
plt.figure(figsize=(8,6))

sns.scatterplot(data=df,
                x="bat_Mat",
                y="bat_Runs")

plt.title("Runs vs Matches")
plt.show()
```

## Bowling vs Economy

```python
plt.figure(figsize=(8,6))

sns.scatterplot(data=df,
                x="bowl_Ave",
                y="bowl_Econ")

plt.title("Bowling Average vs Economy")
plt.show()
```

## Top All rounders 

```python
cols = ['bat_Runs',
        'bat_Ave',
        'bat_SR',
        'bowl_Wkts',
        'bowl_Ave']

sns.pairplot(df[cols])

plt.show()
```

```python
##Detecting outliers in runs

plt.figure(figsize=(8,5))

sns.boxplot(x=df["bat_Runs"])

plt.title("Outliers in Runs")
plt.show()
```
